```python
import pandas as pd
```

```python
dataset = pd.read_csv(r"C:\Users\Sonia Rawat\Downloads\SBI loan dataset\SBI loan dataset\test_Y3wMUE5_7gLdaTN.csv")
```

```python
dataset.head(10)
```

```python
dataset.isnull().sum()
```

```python
dataset["Gender"].fillna(dataset["Gender"].mode()[0],inplace = True)
```

```python
dataset.isnull().sum()
```

```python
en_data = dataset[["Gender","Married"]]
```

```python
pd.get_dummies(en_data)
```

```python
from sklearn.preprocessing import OneHotEncoder
```

```python
ohe = OneHotEncoder()
```

```python
ohe.fit_transform(en_data)
```

```python
ohe.fit_transform(en_data).toarray()
```

```python
ar = ohe.fit_transform(en_data).toarray()
```

```python
pd.DataFrame(ar,columns = ["Gender_Female","Gender_Male","Married_No","Married_Yes"])
```

```python
ohe = OneHotEncoder(drop = "first")
```

```python
ar = ohe.fit_transform(en_data).toarray()
```

```python
pd.DataFrame(ar,columns = ["Gender_Male","Married_Yes"])
```
